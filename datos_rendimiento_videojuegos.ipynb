```python
import pandas as pd

file_1_path = '/content/Gaming_Academic_Performance_updated.csv'
file_2_path = '/content/videojuegos_rendimiento_academico_real.xlsx'

df_gaming_academic = pd.read_csv(file_1_path)
df_videojuegos = pd.read_excel(file_2_path)

print("=== Dataset 1: Gaming Academic Performance ===")
print(f"Shape: {df_gaming_academic.shape}")
print(df_gaming_academic.info())

print("\n=== Dataset 2: Videojuegos Rendimiento Académico ===")
print(f"Shape: {df_videojuegos.shape}")
print(df_videojuegos.info())
```

```python
print("Dataset 1 Preview:")
display(df_gaming_academic.head())
```

```python
print("Dataset 2 Preview:")
display(df_videojuegos.head())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

sns.scatterplot(
    data=df_gaming_academic,
    x='gaming_hours',
    y='grades',
    hue='stress_level',
    alpha=0.6,
    ax=axes[0]
)
axes[0].set_title('Dataset 1: Gaming Hours vs Grades')
axes[0].set_xlabel('Gaming Hours per Day')
axes[0].set_ylabel('Grades (0-100)')

sns.boxplot(
    data=df_videojuegos,
    x='horas_juego_promedio_dia',
    y='promedio_universitario_0_4',
    ax=axes[1]
)
axes[1].set_title('Dataset 2: Daily Game Hours vs University GPA')
axes[1].set_xlabel('Average Daily Game Hours')
axes[1].set_ylabel('GPA (0-4)')

plt.tight_layout()
plt.show()
```

```python
corr_1 = df_gaming_academic[['gaming_hours', 'study_hours', 'sleep_hours', 'attendance', 'grades', 'addiction_score']].corr()

corr_2 = df_videojuegos[['horas_juego_promedio_dia', 'promedio_universitario_0_4', 'nota_secundaria_0_5', 'edad', 'edad_inicio_videojuegos']].corr()

fig, axes = plt.subplots(1, 2, figsize=(18, 7))

sns.heatmap(corr_1, annot=True, cmap='coolwarm', fmt=".2f", ax=axes[0])
axes[0].set_title('Correlation Heatmap: Dataset 1')

sns.heatmap(corr_2, annot=True, cmap='coolwarm', fmt=".2f", ax=axes[1])
axes[1].set_title('Correlation Heatmap: Dataset 2')

plt.tight_layout()
plt.show()
```
