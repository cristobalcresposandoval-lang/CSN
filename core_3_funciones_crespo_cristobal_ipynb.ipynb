```python


ventas = [
    {"fecha": "2024-01-01", "producto": "Laptop", "cantidad": 2, "precio": 850.0},
    {"fecha": "2024-01-01", "producto": "Mouse", "cantidad": 5, "precio": 20.0},
    {"fecha": "2024-01-02", "producto": "Teclado", "cantidad": 3, "precio": 45.0},
    {"fecha": "2024-01-02", "producto": "Mouse", "cantidad": 4, "precio": 20.0},
    {"fecha": "2024-01-03", "producto": "Laptop", "cantidad": 1, "precio": 850.0},
    {"fecha": "2024-01-03", "producto": "Teclado", "cantidad": 2, "precio": 45.0}

]

ingresos_totales = 0

for venta in ventas:
    ingresos_totales += venta["cantidad"] * venta["precio"]


ventas_por_producto = {}

for venta in ventas:
    producto = venta["producto"]

    if producto not in ventas_por_producto:
        ventas_por_producto[producto] = 0

    ventas_por_producto[producto] += venta["cantidad"]

producto_mas_vendido = max(
    ventas_por_producto,
    key=ventas_por_producto.get
)

cantidad_mas_vendida = ventas_por_producto[producto_mas_vendido]

precios_por_producto = {}

for venta in ventas:
    producto = venta["producto"]
    cantidad = venta["cantidad"]
    precio = venta["precio"]

    if producto not in precios_por_producto:
        precios_por_producto[producto] = (0, 0)

    suma_precios, total_unidades = precios_por_producto[producto]

    precios_por_producto[producto] = (
        suma_precios + (precio * cantidad),
        total_unidades + cantidad
    )

precio_promedio_por_producto = {}

for producto, datos in precios_por_producto.items():
    suma_precios, total_unidades = datos
    precio_promedio_por_producto[producto] = (
        suma_precios / total_unidades
    )

ingresos_por_dia = {}

for venta in ventas:
    fecha = venta["fecha"]
    ingreso = venta["cantidad"] * venta["precio"]

    if fecha not in ingresos_por_dia:
        ingresos_por_dia[fecha] = 0

    ingresos_por_dia[fecha] += ingreso

resumen_ventas = {}

for producto in ventas_por_producto:

    cantidad_total = ventas_por_producto[producto]

    ingresos_producto = 0

    for venta in ventas:
        if venta["producto"] == producto:
            ingresos_producto += (
                venta["cantidad"] * venta["precio"]
            )

    resumen_ventas[producto] = {
        "cantidad_total": cantidad_total,
        "ingresos_totales": ingresos_producto,
        "precio_promedio":
            round(precio_promedio_por_producto[producto], 2)
    }
print("=== LISTA DE VENTAS ====")
for venta in ventas:
    print(venta)

print("\n=== INGRESOS TOTALES ====")
print(ingresos_totales)

print("\n=== PRODUCTO MÁS VENDIDO ===")
print(producto_mas_vendido)
print("Cantidad vendida:", cantidad_mas_vendida)

print("\n=== PRECIO PROMEDIO POR PRODUCTO ===")
for producto, promedio in precio_promedio_por_producto.items():
    print(producto, ":", round(promedio, 2))

print("\n=== INGRESOS POR DÍA ===")
for fecha, ingreso in ingresos_por_dia.items():
    print(fecha, ":", ingreso)

print("\n=== RESUMEN DE VENTAS ===")
for producto, datos in resumen_ventas.items():
    print(producto, ":", datos)


with open("reporte_ventas.txt", "w", encoding="utf-8") as archivo:

    archivo.write("REPORTE DE VENTAS")
    archivo.write("===\n")

    archivo.write("LISTA DE VENTAS")
    for venta in ventas:
        archivo.write(str(venta) + "")

    archivo.write("INGRESOS TOTALES")
    archivo.write(str(ingresos_totales) + "")

    archivo.write("PRODUCTO MÁS VENDIDO")
    archivo.write(
        f"{producto_mas_vendido} ({cantidad_mas_vendida} unidades)\n"
    )
    archivo.write("PRECIO PROMEDIO POR PRODUCTO")
    for producto, promedio in precio_promedio_por_producto.items():
        archivo.write(
            f"{producto}: {round(promedio,2)}\n"
        )

    archivo.write("\nINGRESOS POR DÍA\n")
    for fecha, ingreso in ingresos_por_dia.items():
        archivo.write(
            f"{fecha}: {ingreso}\n"
        )
    archivo.write("\nRESUMEN DE VENTAS")
    for producto, datos in resumen_ventas.items():
        archivo.write(
            f"{producto}: {datos}n"
        )

print("Se creó el archivo 'reporte_ventas.txt'")
```
