```python
import numpy as np

def cargar_datos(ruta_archivo):
    """
    Carga los datos del archivo CSV utilizando NumPy
    """
    try:
        # Usamos genfromtxt para cargar los datos numéricos y manejar encabezados
        datos = np.genfromtxt(ruta_archivo, delimiter=',', skip_header=1)
        return datos
    except FileNotFoundError:
        print(f"Error: No se encontró el archivo en la ruta {ruta_archivo}")
        return None


ruta_archivo = 'retail_sales.csv'
print("Función cargar_datos definida correctamente.")
```

```python


datos_ejemplo = np.array([
    [101, 1, 2, 500.0, 1000.0],
    [102, 2, 3, 50.0,  150.0],
    [103, 1, 1, 800.0, 800.0],
    [104, 3, 5, 20.0,  100.0],
    [105, 2, 1, 120.0, 120.0],
    [106, 3, 2, 45.0,   90.0]
])

total_ventas = np.sum(datos_ejemplo[:, 4])
print(f"Total de ventas generales: ${total_ventas}")

promedio_ventas = np.mean(datos_ejemplo[:, 4])
print(f"Promedio de ventas por transacción: ${promedio_ventas:.2f}")

filtro_electronica = datos_ejemplo[:, 1] == 1
ventas_electronica = datos_ejemplo[filtro_electronica]
print("\nVentas filtradas para la categoría 'Electrónica' (ID 1):")
print(ventas_electronica)
```

```python

categorias_nombres = {1: 'Electrónica', 2: 'Ropa', 3: 'Hogar'}
categorias_unicas = np.unique(datos_ejemplo[:, 1])

print("=== 1. TOTAL Y PROMEDIO DE VENTAS POR CATEGORÍA ===")
for cat in categorias_unicas:
    filtro = datos_ejemplo[:, 1] == cat
    ventas_cat = datos_ejemplo[filtro, 4]
    total_cat = np.sum(ventas_cat)
    promedio_cat = np.mean(ventas_cat)
    print(f"Categoría {categorias_nombres[int(cat)]} (ID {int(cat)}): Total = ${total_cat:.2f}, Promedio = ${promedio_cat:.2f}")

print("\n=== 2. CATEGORÍAS CON MAYOR Y MENOR VENTA ===")
totales_por_categoria = np.array([np.sum(datos_ejemplo[datos_ejemplo[:, 1] == cat, 4]) for cat in categorias_unicas])
idx_max = np.argmax(totales_por_categoria)
idx_min = np.argmin(totales_por_categoria)

print(f"Mayor venta: {categorias_nombres[int(categorias_unicas[idx_max])]} (${totales_por_categoria[idx_max]:.2f})")
print(f"Menor venta: {categorias_nombres[int(categorias_unicas[idx_min])]} (${totales_por_categoria[idx_min]:.2f})")

print("\n=== 3. OPERACIONES ARITMÉTICAS ADICIONALES (Impuestos y Descuentos) ===")

impuestos = datos_ejemplo[:, 4] * 0.19
print(f"Impuestos calculados (19% por transacción): {impuestos}")


cantidades = datos_ejemplo[:, 2]
ventas_finales = np.where(cantidades > 2, datos_ejemplo[:, 4] * 0.90, datos_ejemplo[:, 4])
print(f"Venta final aplicando 10% de descuento por volumen: {ventas_finales}")
```
