```python

import seaborn as sns
import matplotlib.pyplot as plt

# 1. Cargar el dataset
df = sns.load_dataset('tips')

# 2. Crear una figura y un eje
fig, ax = plt.subplots(figsize=(12, 6))

# 3. Crear el gráfico de dispersión
sns.scatterplot(
    data=df,
    x='total_bill',
    y='tip',
    hue='time',
    style='smoker',
    s=100,
    ax=ax
)

# 4. Configurar el título y los nombres de los ejes
ax.set_title('Relación entre Total Bill y Tip con Anotaciones y Flechas', fontsize=20)
ax.set_xlabel('Total Bill ($)', fontsize=15)
ax.set_ylabel('Tip ($)', fontsize=15)

# 5. Configurar la leyenda
ax.legend(
    loc='upper right',
    fontsize='large',
    shadow=True,
    title='Detalles'
)

# 6. Añadir cuadrícula
ax.grid(
    True,
    which='major',
    linestyle='--',
    linewidth=0.5,
    color='gray'
)

# 7. Añadir anotación con flecha negra
ax.annotate(
    'Mayor tip en la cena',
    xy=(50.81, 10.0),
    xytext=(48, 9.5),
    arrowprops=dict(facecolor='black', shrink=0.05)
)

# 8. Añadir flecha verde
ax.arrow(
    30, 5,
    10, 3.2,
    head_width=0.8,
    head_length=0.5,
    fc='green',
    ec='green'
)

# 9. Mostrar el gráfico
plt.show()
```
