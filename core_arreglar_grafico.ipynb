```python

import pandas as pd
import matplotlib.pyplot as plt
from io import StringIO

# Datos de ejemplo en formato CSV
data = """fecha,ventas
2023-01-01,200
2023-01-02,220
2023-01-03,210
2023-01-04,205
2023-01-05,215
2023-01-06,225
"""

# Cargar los datos en un DataFrame
df = pd.read_csv(StringIO(data))

# Convertir la columna fecha a formato datetime
df['fecha'] = pd.to_datetime(df['fecha'])

# Mostrar los datos
print(df)

# Gráfico de líneas
plt.figure(figsize=(10, 6))
plt.plot(df['fecha'], df['ventas'], marker='o')
plt.title("Ventas Diarias")
plt.xlabel("Fecha")
plt.ylabel("Ventas")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Gráfico de dispersión
plt.figure(figsize=(10, 6))
plt.scatter(df['fecha'], df['ventas'])
plt.title("Ventas Diarias")
plt.xlabel("Fecha")
plt.ylabel("Ventas")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Histograma de las ventas
plt.figure(figsize=(10, 6))
plt.hist(df['ventas'], bins=5, color='skyblue',
         edgecolor='black')
plt.title("Distribución de Ventas")
plt.xlabel("Ventas")
plt.ylabel("Frecuencia")
plt.show()
```
